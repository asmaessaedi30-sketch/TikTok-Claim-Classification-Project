**Goal:** Determine whether there is a statistically significant difference in viewer counts between verified and unverified TikTok accounts using a two-sample t-test.


```
import pandas as pd
import numpy as np
from scipy import stats
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('../data/tiktok_dataset.csv')
df_clean = df.dropna(subset=['claim_status', 'video_view_count', 'verified_status']).copy()
df_clean.head()
```

## 1. Descriptive Statistics by Verified Status

```
# Compute mean, median, std for view counts grouped by verified status
view_summary = df_clean.groupby('verified_status')['video_view_count'].agg(['count', 'mean', 'median', 'std'])
print(view_summary)
```

## 2. Hypothesis Testing (Two-Sample T-Test)

- **Null Hypothesis ($):** There is no difference in the mean view count for TikTok videos posted by verified accounts and unverified accounts ($\mu_{verified} = \mu_{unverified}$).
- **Alternative Hypothesis ($):** There is a statistically significant difference in the mean view count for TikTok videos posted by verified accounts and unverified accounts ($\mu_{verified} 
eq \mu_{unverified}$).
- **Significance Level ($lpha$):** 0.05


```
verified_views = df_clean[df_clean['verified_status'] == 'verified']['video_view_count']
unverified_views = df_clean[df_clean['verified_status'] == 'not verified']['video_view_count']

# Perform two-sample t-test (equal_var=False for Welch's t-test)
t_stat, p_val = stats.ttest_ind(verified_views, unverified_views, equal_var=False)

print(f'T-statistic: {t_stat:.4f}')
print(f'P-value: {p_val:.4e}')

if p_val < 0.05:
    print('Result: Reject the Null Hypothesis (H0). Statistically significant difference detected!')
else:
    print('Result: Fail to reject the Null Hypothesis (H0).')
```

## 3. Executive Conclusion & Recommendations

- **Finding:** The p-value is far below 0.05, confirming that unverified accounts receive significantly higher view counts than verified accounts on average.
- **Business Implication:** Misinformation and unverified claims have high potential reach. TikTok should prioritize review mechanisms for high-view videos posted by unverified creators.
